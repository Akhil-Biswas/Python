# Python Numbers
There are three numeric types in Python:

1. int
2. float
3. complex

```python
# Example 
x = 1    # int
y = 2.8  # float
z = 1j   # complex
```

```python
# int 
x = 1
y = 35656222554887711
z = -3255522

print(type(x))
print(type(y))
print(type(z))
```

```python
# Float 
x = 1.10
y = 1.0
z = -35.59

print(type(x))
print(type(y))
print(type(z))

# Float can also be scientific numbers with an "e" to indicate the power of 10.
a = 35e3
b = 12E4
c = -87.7e100

print(type(a))
print(type(b))
print(type(c))
```

```python
# Complex
x = 3+5j
y = 5j
z = -5j

print(type(x))
print(type(y))
print(type(z))
```
