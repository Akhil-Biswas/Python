# Slicing Strings
1. You can return a range of characters by using the slice syntax.
2. Specify the <start index> and the <end index>, separated by a <colon>, to return a part of the string.

```python
b = "Hello, World!"
print(b[2:5:2]) # last position not Included <start:end:Step>
```

# Slice From the Start

```python
b = "Hello, World!"
print(b[:5]) # Get the characters from the start to position 5 (not included):
```

#  Slice To the End

```python
b = "Hello, World!"
print(b[2:])
```

# Negative Indexing


```python
b = "Hello, World!"
print(b[-5:-2])
```
