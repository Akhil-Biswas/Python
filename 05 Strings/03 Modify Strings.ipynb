# Upper Case

```python
# The <upper()> method returns the string in upper case:
a = "Hello, World!"
print(a.upper())
```

# Lower Case

```python
# The <lower()> method returns the string in lower case:
a = "Hello, World!"
print(a.lower())
```

# Remove Whitespace

```python
# The <strip()> method removes any whitespace from the ( beginning or the end ):
a = " Hello, World! "
print(a.strip()) # returns "Hello, World!"
```

# Replace String

```python
# The <replace()> method replaces a string with another string:

a = "Hello, World!"
print(a.replace("H", "J"))
```

```python
a = "hello, World!"
print(a.replace("H", "J")) # case sansitive
```

```python
a = "Hello, World!"
print(a.replace("l", "J")) # it replace all 
```

# Split String

```python
# The split() method splits the string into substrings if it finds instances of the separator:

a = "Hello, World!"
print(a.split(",")) # Retrune velue in List Type
```
